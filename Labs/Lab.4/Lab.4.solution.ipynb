1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class Counter:
    def __init__(self, max_value):
        self.max_value = max_value
        self.value = 0

    def increment(self):
        if self.value >= self.max_value:
            print(f"Error: Cannot increment beyond maximum value of {self.max_value}.")
            return
        self.value += 1

    def reset(self):
        self.value = 0

    def __repr__(self):
        return f"Counter(value={self.value}, max_value={self.max_value})"

c = Counter(3)
c.increment()  
c.increment()  
c.increment()  
c.increment()  
print(c)   
c.reset()
print(c)   
```

8. Add a function to the object classes, including the base, that tests if a given set of $x$ and $y$ coordinates are inside of the object. You'll have to think through how to determine if a set of coordinates are inside an object for each object type.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    def __init__(self, x, y):
        self._x = x
        self._y = y

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    @abstractmethod
    def perimeter_points(self, num_points=16):
        pass

    @abstractmethod
    def contains_point(self, px, py):
        pass

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def get_position(self):
        return (self._x, self._y)

    @staticmethod
    def _points_on_polygon(vertices, num_points):
        n = len(vertices)
        edges = []
        total_length = 0.0
        for i in range(n):
            p1 = vertices[i]
            p2 = vertices[(i + 1) % n]
            length = math.dist(p1, p2)
            edges.append((p1, p2, length))
            total_length += length

        points = []
        for i in range(num_points):
            target = total_length * i / num_points
            accumulated = 0.0
            for p1, p2, length in edges:
                if length > 0 and accumulated + length >= target:
                    t = (target - accumulated) / length
                    px = p1[0] + t * (p2[0] - p1[0])
                    py = p1[1] + t * (p2[1] - p1[1])
                    points.append((px, py))
                    break
                accumulated += length
            else:
                points.append(vertices[0])
        return points

    @staticmethod
    def _point_in_polygon(px, py, vertices):
        n = len(vertices)
        inside = False
        x1, y1 = vertices[0]
        for i in range(1, n + 1):
            x2, y2 = vertices[i % n]
            if ((y1 > py) != (y2 > py)):
                x_intersect = (x2 - x1) * (py - y1) / (y2 - y1) + x1
                if px < x_intersect:
                    inside = not inside
            x1, y1 = x2, y2
        return inside

    def __repr__(self):
        return f"{self.__class__.__name__}(area={self.area()}, perimeter={self.perimeter()})"

class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_corner(self):
        return self.get_position()

    def _vertices(self):
        x, y = self._x, self._y
        return [
            (x, y),
            (x + self.__length, y),
            (x + self.__length, y + self.__width),
            (x, y + self.__width),
        ]

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        return self._points_on_polygon(self._vertices(), num_points)

    def contains_point(self, px, py):
        x, y = self._x, self._y
        return (x <= px <= x + self.__length) and (y <= py <= y + self.__width)

class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_center(self):
        return self.get_position()

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        points = []
        for i in range(num_points):
            angle = 2 * math.pi * i / num_points
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def contains_point(self, px, py):
        return math.dist((px, py), (self._x, self._y)) <= self.__radius

class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def area(self):
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self.__side_a) * (s - self.__side_b) * (s - self.__side_c))

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def get_vertex(self):
        return self.get_position()

    def _vertices(self):
        x, y = self._x, self._y
        a, b, c = self.__side_a, self.__side_b, self.__side_c

        v1 = (x, y)
        v2 = (x + a, y)

        cos_angle = (a ** 2 + c ** 2 - b ** 2) / (2 * a * c)
        cos_angle = max(-1.0, min(1.0, cos_angle))
        angle = math.acos(cos_angle)

        v3 = (x + c * math.cos(angle), y + c * math.sin(angle))
        return [v1, v2, v3]

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        return self._points_on_polygon(self._vertices(), num_points)

    def contains_point(self, px, py):
        v1, v2, v3 = self._vertices()
        x1, y1 = v1
        x2, y2 = v2
        x3, y3 = v3

        denom = (y2 - y3) * (x1 - x3) + (x3 - x2) * (y1 - y3)
        if denom == 0:
            return False  

        w1 = ((y2 - y3) * (px - x3) + (x3 - x2) * (py - y3)) / denom
        w2 = ((y3 - y1) * (px - x3) + (x1 - x3) * (py - y3)) / denom
        w3 = 1 - w1 - w2

        return w1 >= 0 and w2 >= 0 and w3 >= 0

r = Rectangle(4, 2, 0, 0)
print(r.contains_point(2, 1))   
print(r.contains_point(5, 1))   
print(r.contains_point(4, 2))    
c = Circle(2, 0, 0)
print(c.contains_point(1, 1))    
print(c.contains_point(2, 2))   
t = Triangle(3, 4, 5, 0, 0)
print(t.contains_point(1, 1))    
print(t.contains_point(4, 4))   
```

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    def __init__(self, x, y):
        self._x = x
        self._y = y

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    @abstractmethod
    def perimeter_points(self, num_points=16):
        pass

    @abstractmethod
    def contains_point(self, px, py):
        pass

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def get_position(self):
        return (self._x, self._y)

    def overlaps(self, other):
        if other.contains_point(self._x, self._y):
            return True
        if self.contains_point(other.get_x(), other.get_y()):
            return True

        for (px, py) in self.perimeter_points():
            if other.contains_point(px, py):
                return True
                
        for (px, py) in other.perimeter_points():
            if self.contains_point(px, py):
                return True

        return False

    @staticmethod
    def _points_on_polygon(vertices, num_points):
        n = len(vertices)
        edges = []
        total_length = 0.0
        for i in range(n):
            p1 = vertices[i]
            p2 = vertices[(i + 1) % n]
            length = math.dist(p1, p2)
            edges.append((p1, p2, length))
            total_length += length

        points = []
        for i in range(num_points):
            target = total_length * i / num_points
            accumulated = 0.0
            for p1, p2, length in edges:
                if length > 0 and accumulated + length >= target:
                    t = (target - accumulated) / length
                    px = p1[0] + t * (p2[0] - p1[0])
                    py = p1[1] + t * (p2[1] - p1[1])
                    points.append((px, py))
                    break
                accumulated += length
            else:
                points.append(vertices[0])
        return points

    @staticmethod
    def _point_in_polygon(px, py, vertices):
        n = len(vertices)
        inside = False
        x1, y1 = vertices[0]
        for i in range(1, n + 1):
            x2, y2 = vertices[i % n]
            if ((y1 > py) != (y2 > py)):
                x_intersect = (x2 - x1) * (py - y1) / (y2 - y1) + x1
                if px < x_intersect:
                    inside = not inside
            x1, y1 = x2, y2
        return inside

    def __repr__(self):
        return f"{self.__class__.__name__}(area={self.area()}, perimeter={self.perimeter()})"


class Rectangle(Shape):
    def __init__(self, length, width, x, y):
        super().__init__(x, y)
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_corner(self):
        return self.get_position()

    def _vertices(self):
        x, y = self._x, self._y
        return [
            (x, y),
            (x + self.__length, y),
            (x + self.__length, y + self.__width),
            (x, y + self.__width),
        ]

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        return self._points_on_polygon(self._vertices(), num_points)

    def contains_point(self, px, py):
        x, y = self._x, self._y
        return (x <= px <= x + self.__length) and (y <= py <= y + self.__width)


class Circle(Shape):
    def __init__(self, radius, x, y):
        super().__init__(x, y)
        self.__radius = radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_center(self):
        return self.get_position()

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        points = []
        for i in range(num_points):
            angle = 2 * math.pi * i / num_points
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def contains_point(self, px, py):
        return math.dist((px, py), (self._x, self._y)) <= self.__radius


class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y):
        super().__init__(x, y)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def area(self):
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self.__side_a) * (s - self.__side_b) * (s - self.__side_c))

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def get_vertex(self):
        return self.get_position()

    def _vertices(self):
        x, y = self._x, self._y
        a, b, c = self.__side_a, self.__side_b, self.__side_c

        v1 = (x, y)
        v2 = (x + a, y)

        cos_angle = (a ** 2 + c ** 2 - b ** 2) / (2 * a * c)
        cos_angle = max(-1.0, min(1.0, cos_angle))
        angle = math.acos(cos_angle)

        v3 = (x + c * math.cos(angle), y + c * math.sin(angle))
        return [v1, v2, v3]

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        return self._points_on_polygon(self._vertices(), num_points)

    def contains_point(self, px, py):
        v1, v2, v3 = self._vertices()
        x1, y1 = v1
        x2, y2 = v2
        x3, y3 = v3

        denom = (y2 - y3) * (x1 - x3) + (x3 - x2) * (y1 - y3)
        if denom == 0:
            return False

        w1 = ((y2 - y3) * (px - x3) + (x3 - x2) * (py - y3)) / denom
        w2 = ((y3 - y1) * (px - x3) + (x1 - x3) * (py - y3)) / denom
        w3 = 1 - w1 - w2

        return w1 >= 0 and w2 >= 0 and w3 >= 0

r1 = Rectangle(4, 2, 0, 0)
r2 = Rectangle(4, 2, 3, 1)   
r3 = Rectangle(4, 2, 100, 100) 
print(r1.overlaps(r2))   
print(r1.overlaps(r3))   
c1 = Circle(2, 0, 0)
c2 = Circle(2, 3, 0)    
c3 = Circle(2, 10, 10)   
print(c1.overlaps(c2))   
print(c1.overlaps(c3))   
t = Triangle(3, 4, 5, 0, 0)
print(t.overlaps(r1))   
```

```python
import math
from abc import ABC, abstractmethod

class Counter:
    def __init__(self, max_value):
        self.__max_value = max_value
        self.__value = 0

    def increment(self):
        if self.__value >= self.__max_value:
            print(f"Error: Cannot increment beyond maximum value of {self.__max_value}.")
            return
        self.__value += 1

    def reset(self):
        self.__value = 0

    def get_value(self):
        return self.__value

    def get_max_value(self):
        return self.__max_value

    def is_at_max(self):
        return self.__value >= self.__max_value

    def __repr__(self):
        return f"Counter(value={self.__value}, max_value={self.__max_value})"

class Canvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.data = [[' '] * width for i in range(height)]

    def set_pixel(self, row, col, char='*'):
        self.data[row][col] = char

    def get_pixel(self, row, col):
        return self.data[row][col]

    def clear_canvas(self):
        self.data = [[' '] * self.width for i in range(self.height)]

    def v_line(self, x, y, h, **kargs):
        for i in range(x, x + h):
            self.set_pixel(i, y, **kargs)

    def h_line(self, x, y, w, **kargs):
        for i in range(y, y + w):
            self.set_pixel(x, i, **kargs)

    def line(self, x1, y1, x2, y2, **kargs):
        slope = (x2 - x1) / (y2 - y1)
        for y in range(y1, y2):
            x = x1 + int(slope * (y - y1))
            self.set_pixel(x, y, **kargs)

    def display(self):
        print("\n".join(["".join(row) for row in self.data]))

class Shape(ABC):
    def __init__(self, x, y, name=""):
        self._x = x
        self._y = y
        self.name = name

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

    @abstractmethod
    def perimeter_points(self, num_points=16):
        pass

    @abstractmethod
    def contains_point(self, px, py):
        pass

    @abstractmethod
    def paint(self, canvas):
        pass

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def get_position(self):
        return (self._x, self._y)

    def move(self, dx, dy):
        self._x += dx
        self._y += dy

    def overlaps(self, other):
        if other.contains_point(self._x, self._y):
            return True
        if self.contains_point(other.get_x(), other.get_y()):
            return True
        for (px, py) in self.perimeter_points():
            if other.contains_point(px, py):
                return True
        for (px, py) in other.perimeter_points():
            if self.contains_point(px, py):
                return True
        return False

    @staticmethod
    def _plot_point(canvas, x, y, char='*'):
        row = int(round(y))
        col = int(round(x))
        if 0 <= row < canvas.height and 0 <= col < canvas.width:
            canvas.set_pixel(row, col, char)

    @classmethod
    def _draw_segment(cls, canvas, p1, p2, char='*'):
        x1, y1 = p1
        x2, y2 = p2
        distance = math.dist(p1, p2)
        steps = max(1, int(distance * 2))
        for i in range(steps + 1):
            t = i / steps
            x = x1 + t * (x2 - x1)
            y = y1 + t * (y2 - y1)
            cls._plot_point(canvas, x, y, char)

    @staticmethod
    def _points_on_polygon(vertices, num_points):
        n = len(vertices)
        edges = []
        total_length = 0.0
        for i in range(n):
            p1 = vertices[i]
            p2 = vertices[(i + 1) % n]
            length = math.dist(p1, p2)
            edges.append((p1, p2, length))
            total_length += length

        points = []
        for i in range(num_points):
            target = total_length * i / num_points
            accumulated = 0.0
            for p1, p2, length in edges:
                if length > 0 and accumulated + length >= target:
                    t = (target - accumulated) / length
                    px = p1[0] + t * (p2[0] - p1[0])
                    py = p1[1] + t * (p2[1] - p1[1])
                    points.append((px, py))
                    break
                accumulated += length
            else:
                points.append(vertices[0])
        return points

    def __str__(self):
        return f"{self.__class__.__name__}(area={self.area()}, perimeter={self.perimeter()})"

class Rectangle(Shape):
    def __init__(self, length, width, x, y, name=""):
        super().__init__(x, y, name)
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def get_corner(self):
        return self.get_position()

    def _vertices(self):
        x, y = self._x, self._y
        return [
            (x, y),
            (x + self.__length, y),
            (x + self.__length, y + self.__width),
            (x, y + self.__width),
        ]

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        return self._points_on_polygon(self._vertices(), num_points)

    def contains_point(self, px, py):
        x, y = self._x, self._y
        return (x <= px <= x + self.__length) and (y <= py <= y + self.__width)

    def paint(self, canvas):
        vertices = self._vertices()
        for i in range(len(vertices)):
            self._draw_segment(canvas, vertices[i], vertices[(i + 1) % len(vertices)])

    def __repr__(self):
        return (f"Rectangle({self.__length!r}, {self.__width!r}, "
                f"{self._x!r}, {self._y!r}, {self.name!r})")

class Circle(Shape):
    def __init__(self, radius, x, y, name=""):
        super().__init__(x, y, name)
        self.__radius = radius

    def area(self):
        return math.pi * self.__radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.__radius

    def get_radius(self):
        return self.__radius

    def get_center(self):
        return self.get_position()

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        points = []
        for i in range(num_points):
            angle = 2 * math.pi * i / num_points
            px = self._x + self.__radius * math.cos(angle)
            py = self._y + self.__radius * math.sin(angle)
            points.append((px, py))
        return points

    def contains_point(self, px, py):
        return math.dist((px, py), (self._x, self._y)) <= self.__radius

    def paint(self, canvas, num_samples=100):
        for i in range(num_samples):
            angle = 2 * math.pi * i / num_samples
            x = self._x + self.__radius * math.cos(angle)
            y = self._y + self.__radius * math.sin(angle)
            self._plot_point(canvas, x, y)

    def __repr__(self):
        return f"Circle({self.__radius!r}, {self._x!r}, {self._y!r}, {self.name!r})"

class Triangle(Shape):
    def __init__(self, side_a, side_b, side_c, x, y, name=""):
        super().__init__(x, y, name)
        self.__side_a = side_a
        self.__side_b = side_b
        self.__side_c = side_c

    def area(self):
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self.__side_a) * (s - self.__side_b) * (s - self.__side_c))

    def perimeter(self):
        return self.__side_a + self.__side_b + self.__side_c

    def get_side_a(self):
        return self.__side_a

    def get_side_b(self):
        return self.__side_b

    def get_side_c(self):
        return self.__side_c

    def get_vertex(self):
        return self.get_position()

    def _vertices(self):
        x, y = self._x, self._y
        a, b, c = self.__side_a, self.__side_b, self.__side_c

        v1 = (x, y)
        v2 = (x + a, y)

        cos_angle = (a ** 2 + c ** 2 - b ** 2) / (2 * a * c)
        cos_angle = max(-1.0, min(1.0, cos_angle))
        angle = math.acos(cos_angle)

        v3 = (x + c * math.cos(angle), y + c * math.sin(angle))
        return [v1, v2, v3]

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        return self._points_on_polygon(self._vertices(), num_points)

    def contains_point(self, px, py):
        v1, v2, v3 = self._vertices()
        x1, y1 = v1
        x2, y2 = v2
        x3, y3 = v3

        denom = (y2 - y3) * (x1 - x3) + (x3 - x2) * (y1 - y3)
        if denom == 0:
            return False

        w1 = ((y2 - y3) * (px - x3) + (x3 - x2) * (py - y3)) / denom
        w2 = ((y3 - y1) * (px - x3) + (x1 - x3) * (py - y3)) / denom
        w3 = 1 - w1 - w2

        return w1 >= 0 and w2 >= 0 and w3 >= 0

    def paint(self, canvas):
        vertices = self._vertices()
        for i in range(len(vertices)):
            self._draw_segment(canvas, vertices[i], vertices[(i + 1) % len(vertices)])

    def __repr__(self):
        return (f"Triangle({self.__side_a!r}, {self.__side_b!r}, {self.__side_c!r}, "
                f"{self._x!r}, {self._y!r}, {self.name!r})")

class CompoundShape(Shape):
    def __init__(self, shapes, x=None, y=None, name=""):
        self.shapes = list(shapes)

        if x is None or y is None:
            if self.shapes:
                avg_x = sum(s.get_x() for s in self.shapes) / len(self.shapes)
                avg_y = sum(s.get_y() for s in self.shapes) / len(self.shapes)
            else:
                avg_x, avg_y = 0, 0
            x = avg_x if x is None else x
            y = avg_y if y is None else y

        super().__init__(x, y, name)

    def add_shape(self, shape):
        self.shapes.append(shape)

    def move(self, dx, dy):
        super().move(dx, dy)
        for s in self.shapes:
            s.move(dx, dy)

    def area(self):
        return sum(s.area() for s in self.shapes)

    def perimeter(self):
        return sum(s.perimeter() for s in self.shapes)

    def perimeter_points(self, num_points=16):
        num_points = min(num_points, 16)
        if not self.shapes:
            return []
        points = []
        per_shape = max(1, num_points // len(self.shapes))
        for s in self.shapes:
            points.extend(s.perimeter_points(per_shape))
        return points[:num_points]

    def contains_point(self, px, py):
        return any(s.contains_point(px, py) for s in self.shapes)

    def paint(self, canvas):
        for s in self.shapes:
            s.paint(canvas)

    def __repr__(self):
        shapes_repr = "[" + ", ".join(repr(s) for s in self.shapes) + "]"
        return f"CompoundShape({shapes_repr}, {self._x!r}, {self._y!r}, {self.name!r})"

class RasterDrawing:
    def __init__(self):
        self.shapes = dict()
        self.shape_names = list()

    def add_shape(self, shape):
        if shape.name == "":
            shape.name = self.assign_name()

        self.shapes[shape.name] = shape
        self.shape_names.append(shape.name)

    def remove_shape(self, name):
        if name in self.shapes:
            del self.shapes[name]
            self.shape_names.remove(name)

    def update(self, canvas):
        canvas.clear_canvas()
        self.paint(canvas)

    def paint(self, canvas):
        for shape_name in self.shape_names:
            self.shapes[shape_name].paint(canvas)

    def assign_name(self):
        name_base = "shape"
        name = name_base + "_0"

        i = 1
        while name in self.shapes:
            name = name_base + "_" + str(i)
            i += 1

        return name

    def __repr__(self):
        shape_reprs = ", ".join(repr(self.shapes[name]) for name in self.shape_names)
        return f"build_raster_drawing([{shape_reprs}])"

    def save(self, filename):
        f = open(filename, "w")
        f.write(repr(self))
        f.close()

def build_raster_drawing(shapes):
    rd = RasterDrawing()
    for s in shapes:
        rd.add_shape(s)
    return rd

def raster_drawing_loader(filename):
    f = open(filename, "r")
    tmp = eval(f.read())
    f.close()
    return tmp

if __name__ == "__main__":
    print("--- Counter demo ---")
    counter = Counter(3)
    counter.increment()
    counter.increment()
    counter.increment()
    counter.increment()  
    print(counter, "| at max:", counter.is_at_max())

    print("\n--- Drawing demo ---")
    canvas = Canvas(40, 40)

    rectangle = Rectangle(length=15, width=10, x=2, y=2)
    circle = Circle(radius=6, x=30, y=10)
    triangle = Triangle(side_a=10, side_b=10, side_c=12, x=5, y=20)

    small_circle = Circle(radius=3, x=30, y=30)
    small_rectangle = Rectangle(length=6, width=4, x=20, y=28)
    compound = CompoundShape([small_circle, small_rectangle])

    for shape in (rectangle, circle, triangle, compound):
        shape.paint(canvas)

    canvas.display()

    print("\nRectangle area/perimeter:", rectangle.area(), rectangle.perimeter())
    print("Circle area/perimeter:", round(circle.area(), 2), round(circle.perimeter(), 2))
    print("Triangle area/perimeter:", round(triangle.area(), 2), triangle.perimeter())
    print("Compound area/perimeter:", round(compound.area(), 2), round(compound.perimeter(), 2))
    print("Compound contains (30, 30)?", compound.contains_point(30, 30))
    print("Rectangle overlaps Triangle?", rectangle.overlaps(triangle))

    print("\n--- RasterDrawing demo ---")
    rd = RasterDrawing()
    rd.add_shape(Rectangle(length=8, width=6, x=2, y=2))
    rd.add_shape(Circle(radius=5, x=25, y=10))
    rd.add_shape(Triangle(side_a=8, side_b=8, side_c=10, x=5, y=20))

    drawing_canvas = Canvas(40, 40)
    rd.paint(drawing_canvas)
    print("Initial drawing:")
    drawing_canvas.display()

    rd.shapes["shape_0"].move(15, 0)
    rd.add_shape(Circle(radius=3, x=35, y=35, name="extra_circle"))

    rd.update(drawing_canvas)
    print("\nAfter moving shape_0 and adding a circle:")
    drawing_canvas.display()

    print("\n--- Save/load demo ---")
    print("repr(rd):", repr(rd))

    rd.save("drawing.raster")
    with open("drawing.raster") as f:
        print("\nSaved file contents:", f.read())

    rd_reloaded = raster_drawing_loader("drawing.raster")

    reloaded_canvas = Canvas(40, 40)
    rd_reloaded.paint(reloaded_canvas)
    print("Reloaded drawing (should match the drawing above):")
    reloaded_canvas.display()

    print("\nSame object?", rd_reloaded is rd)
    print("Same shape names?", rd_reloaded.shape_names == rd.shape_names)
    print(
        "shape_0 position matches?",
        rd_reloaded.shapes["shape_0"].get_position() == rd.shapes["shape_0"].get_position(),
    )
```
