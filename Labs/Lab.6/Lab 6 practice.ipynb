The two versions don't detect the same spikes. With factor = 2, the median flags IDs 5, 8, and 12, while the mean flags only 5 and 12, missing ID 8. The reason is that the earlier spike (135 at ID 5) is still in the window when ID 8 is checked. It pulls the mean of the last three readings (135, 39, 45) up to 73, so the threshold becomes 146 and 120 falls below it. The median of those same readings is 45, so the threshold is only 90 and 120 is flagged. The mean uses every value in the window, so one extreme reading can inflate the baseline and hide the next spike, which is called masking. The median depends only on the middle value, so it is robust to outliers, which makes it the better baseline for spike detection.

```python
from statistics import median

readings = [
    {"id": 1, "amount": 40}, {"id": 2, "amount": 42},
    {"id": 3, "amount": 38}, {"id": 4, "amount": 41},
    {"id": 5, "amount": 135}, {"id": 6, "amount": 39},
    {"id": 7, "amount": 45}, {"id": 8, "amount": 120},
    {"id": 9, "amount": 47}, {"id": 10, "amount": None},
    {"id": 11, "amount": 50}, {"id": 12, "amount": 210},
]

def flag_spikes(rows, factor=2.0, window=3):
    history = []
    flagged = []
    for row in rows:
        value = row["amount"]
        if value is None:
            continue
        if len(history) >= window:
            baseline = median(history[-window:])
            if value > factor * baseline:
                flagged.append(row["id"])
        history.append(value)
    return flagged

print("Flagged IDs:", flag_spikes(readings))
```
